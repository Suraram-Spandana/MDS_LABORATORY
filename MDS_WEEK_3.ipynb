```python
import matplotlib.pyplot as plt
import numpy as np
import cv2

# -----------------------------
# Load Image
# -----------------------------
img = cv2.imread("cat.jpg")      # OpenCV reads image in BGR format

# Convert BGR to RGB
rgb_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert to NumPy array (already a NumPy array from imread, but good for explicit type)
rgb_img = np.array(rgb_img)

# Image dimensions
height, width, channels = rgb_img.shape

print("Image converted to NumPy array")
print("Image Shape :", rgb_img.shape)
print("Height :", height)
print("Width :", width)
print("Channels :", channels)
print("Data Type :", rgb_img.dtype)

# -----------------------------
# Crop Entire Cat
# -----------------------------
cat_crop = rgb_img[
    int(height*0.15):int(height*0.95),
    int(width*0.15):int(width*0.85)
]

print("\nCropped Cat Shape :", cat_crop.shape)

# -----------------------------
# Crop ONLY the Eye Region
# -----------------------------
eyes_crop = rgb_img[
    int(height*0.28):int(height*0.36),   # Smaller height
    int(width*0.28):int(width*0.72)      # Eye width only
]

print("\nEye Region Shape :", eyes_crop.shape)
print("Eye Height :", eyes_crop.shape[0])
print("Eye Width :", eyes_crop.shape[1])

# -----------------------------
# Left Eye
# -----------------------------
left_eye = rgb_img[
    int(height*0.29):int(height*0.35),
    int(width*0.30):int(width*0.43)
]

# -----------------------------
# Right Eye
# -----------------------------
right_eye = rgb_img[
    int(height*0.29):int(height*0.35),
    int(width*0.57):int(width*0.70)
]

print("\nLeft Eye Shape :", left_eye.shape)
print("Left Eye Height :", left_eye.shape[0])
print("Left Eye Width :", left_eye.shape[1])

print("\nRight Eye Shape :", right_eye.shape)
print("Right Eye Height :", right_eye.shape[0])
print("Right Eye Width :", right_eye.shape[1])

# -----------------------------
# Flip Operations
# -----------------------------
horizontal_flip = np.fliplr(cat_crop)
vertical_flip = np.flipud(cat_crop)
eyes_flip = np.fliplr(eyes_crop)

# -----------------------------
# Rotation
# -----------------------------
rotate_90 = np.rot90(cat_crop)
rotate_270 = np.rot90(cat_crop, 3)

# -----------------------------
# Color Manipulation
# -----------------------------
# Create a copy of the original RGB image to modify its colors
modified_color_img = rgb_img.copy()

# Example: Increase the red channel by a fixed value
# Pixel values are typically 0-255 for uint8 images.
# Ensure values stay within 0-255 range using np.clip.
# You can change '50' to any value between -255 and 255.
# Experiment with different channels:
#   0: Red channel
#   1: Green channel
#   2: Blue channel
modified_color_img[:, :, 0] = np.clip(modified_color_img[:, :, 0].astype(np.int32) + 50, 0, 255).astype(np.uint8)
print(f"\nModified color image shape: {modified_color_img.shape}")

# -----------------------------
# Display Results
# -----------------------------
plt.figure(figsize=(18,18)) # Increased figure size to accommodate more subplots

# Original
plt.subplot(3,3,1)
plt.imshow(rgb_img)
plt.title("Original RGB Image")
plt.axis("off")

# Cropped Cat
plt.subplot(3,3,2)
plt.imshow(cat_crop)
plt.title("Cropped Cat")
plt.axis("off")

# Horizontal Flip
plt.subplot(3,3,3)
plt.imshow(horizontal_flip)
plt.title("Horizontal Flip")
plt.axis("off")

# Vertical Flip
plt.subplot(3,3,4)
plt.imshow(vertical_flip)
plt.title("Vertical Flip")
plt.axis("off")

# Eye Crop
plt.subplot(3,3,5)
plt.imshow(eyes_crop)
plt.title("Cropped Eyes")
plt.axis("off")

# Flipped Eyes
plt.subplot(3,3,6)
plt.imshow(eyes_flip)
plt.title("Flipped Eyes")
plt.axis("off")

# Rotate 90°
plt.subplot(3,3,7)
plt.imshow(rotate_90)
plt.title("90° Rotation")
plt.axis("off")

# Rotate 270°
plt.subplot(3,3,8)
plt.imshow(rotate_270)
plt.title("270° Rotation")
plt.axis("off")

# Color Modified Image
plt.subplot(3,3,9)
plt.imshow(modified_color_img)
plt.title("Red Channel Increased")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import numpy as np

# -----------------------------
# Load CSV File
# -----------------------------
file_path = "DataSet.csv"

# Numeric columns:
# RollNo, Maths, Physics, Chemistry, English, Computer
numeric_cols = [0, 2, 3, 4, 5, 6]

data = np.genfromtxt(
    file_path,
    delimiter=",",
    skip_header=1,
    usecols=numeric_cols,
    dtype=float
)

print("="*60)
print("DATASET INFORMATION")
print("="*60)

print("Shape :", data.shape)
print("Dimensions :", data.ndim)
print("Size :", data.size)
print("Data Type :", data.dtype)

print("\nFirst 5 Rows")
print(data[:5])

# ======================================================
# Statistical Operations
# ======================================================

print("\n" + "="*60)
print("STATISTICAL OPERATIONS")
print("="*60)

print("\nMean")
print(np.mean(data,axis=0))

print("\nMedian")
print(np.median(data,axis=0))

print("\nStandard Deviation")
print(np.std(data,axis=0))

print("\nVariance")
print(np.var(data,axis=0))

print("\nMinimum")
print(np.min(data,axis=0))

print("\nMaximum")
print(np.max(data,axis=0))

print("\nRange (Max-Min)")
print(np.ptp(data,axis=0))

print("\n25th Percentile")
print(np.percentile(data,25,axis=0))

print("\n50th Percentile")
print(np.percentile(data,50,axis=0))

print("\n75th Percentile")
print(np.percentile(data,75,axis=0))

print("\nColumn Sum")
print(np.sum(data,axis=0))

print("\nRow Sum (First 5)")
print(np.sum(data,axis=1)[:5])

print("\nColumn Product")
print(np.prod(data,axis=0))

# ======================================================
# Comparison Operations
# ======================================================

print("\n" + "="*60)
print("COMPARISON OPERATIONS")
print("="*60)

print("\nElements > 80")
print(data>80)

print("\nNumber of elements >80")
print(np.sum(data>80))

print("\nElements <70")
print(np.sum(data<70))

print("\nMaximum value index in each column")
print(np.argmax(data,axis=0))

print("\nMinimum value index in each column")
print(np.argmin(data,axis=0))

print("\nRows where Maths > 80")
maths = data[:,1]
print(data[maths>80])

# ======================================================
# Array Operations
# ======================================================

print("\n" + "="*60)
print("ARRAY OPERATIONS")
print("="*60)

print("\nTranspose")
print(data.T)

print("\nSorted Data (Column-wise)")
print(np.sort(data,axis=0))

print("\nUnique Roll Numbers")
print(np.unique(data[:,0]))

print("\nColumn Maximum")
print(np.max(data,axis=0))

print("\nColumn Minimum")
print(np.min(data,axis=0))

print("\nColumn Average")
print(np.mean(data,axis=0))

print("\nRow Average (First 5)")
print(np.mean(data,axis=1)[:5])
```
