```python
import time
import numpy as np

# Define a large size for comparison
size = 10**6

# --- Python List ---
print(f"Comparing Python list vs. NumPy array for {size} elements:")

# Create a Python list
start_time = time.time()
python_list = list(range(size))
end_time = time.time()
print(f"Python List creation time: {end_time - start_time:.6f} seconds")

# Perform an operation (e.g., element-wise addition)
start_time = time.time()
python_list_result = [x + 1 for x in python_list]
end_time = time.time()
print(f"Python List element-wise addition time: {end_time - start_time:.6f} seconds")

# --- NumPy Array ---

# Create a NumPy array
start_time = time.time()
numpy_array = np.arange(size)
end_time = time.time()
print(f"NumPy Array creation time: {end_time - start_time:.6f} seconds")

# Perform an operation (e.g., element-wise addition)
start_time = time.time()
numpy_array_result = numpy_array + 1
end_time = time.time()
print(f"NumPy Array element-wise addition time: {end_time - start_time:.6f} seconds")
```

```python
import sys

# --- Memory Comparison ---
print(f"\nComparing memory usage for {size} elements:")

# Memory usage of Python list
# sys.getsizeof() only reports the size of the list object itself, not the elements it contains.
# To get a more accurate total, we sum the size of the list object and each element.
python_list_memory = sys.getsizeof(python_list) + sum(sys.getsizeof(item) for item in python_list)
print(f"Python List memory usage: {python_list_memory / (1024*1024):.4f} MB")

# Memory usage of NumPy array
# .nbytes gives the total bytes consumed by the array elements.
numpy_array_memory = numpy_array.nbytes
print(f"NumPy Array memory usage: {numpy_array_memory / (1024*1024):.4f} MB")

print("\nSummary: NumPy arrays are generally more memory-efficient and significantly faster for numerical operations on large datasets compared to Python lists.")
```

```python
# Create a Python list of lists to represent a 2D array
python_2d_list = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Python 2D List:")
print(python_2d_list)
```

```python
import numpy as np

# Convert the Python list of lists to a NumPy 2D array
numpy_2d_array = np.array(python_2d_list)

print("\nNumPy 2D Array:")
print(numpy_2d_array)
print("Shape of NumPy 2D Array:", numpy_2d_array.shape)
print("Dimensions of NumPy 2D Array:", numpy_2d_array.ndim)
```
